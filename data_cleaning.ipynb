### Project-defined temperature and humidity thresholds

The values used below (temperature above **10°C** and humidity above **85%**) are fixed, project-defined thresholds for categorizing sensor readings as `HIGH` or `NORMAL`. They are used for this project's analysis only and should **not** be interpreted as scientifically validated food-spoilage or food-safety limits.

```python
df['temperature_level'] = np.where(
    df['temperature_c'] > 10,
    'HIGH',
    'NORMAL'
)

print(df['temperature_level'].value_counts())
```

```python
df['humidity_level'] = np.where(
    df['humidity_pct'] > 85,
    'HIGH',
    'NORMAL'
)

print(df['humidity_level'].value_counts())
```

### Outlier review and handling decision

The IQR method identifies **9 vibration outliers** in this dataset. These records are retained in the cleaned dataset rather than removed or capped because unusually high vibration readings may represent genuine transport or sensor alerts. They should be reviewed as operational events, not automatically treated as data errors.

```python
columns = [
    'quantity_kg',
    'temperature_c',
    'humidity_pct',
    'vibration_g'
]

for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print(col, "Outliers:", len(outliers))
```

```python
Q1 = df['vibration_g'].quantile(0.25)
Q3 = df['vibration_g'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

vibration_outliers = df[
    (df['vibration_g'] < lower) |
    (df['vibration_g'] > upper)
]

print(vibration_outliers[
    ['event_id', 'vibration_g', 'sensor_status']
])
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scale_columns = [
    'quantity_kg',
    'temperature_c',
    'humidity_pct',
    'vibration_g'
]

df[[col + '_scaled' for col in scale_columns]] = (
    scaler.fit_transform(df[scale_columns])
)

print(df.head())
```
